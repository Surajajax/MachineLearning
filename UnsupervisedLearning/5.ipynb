```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('data.csv')
data
```

```python
data.columns
data.info()
data.describe()
```

```python
data = data.drop(['id', 'Unnamed: 32'], axis=1)

# Convert diagnosis to binary
# B = 0 (normal)
# M = 1 (anomaly)
data['diagnosis'] = data['diagnosis'].map({'B': 0, 'M': 1})
```

```python
X = data.drop('diagnosis', axis=1)
y = data['diagnosis']
```

```python
X_train = X[y == 0].copy()
```

```python
mu = X_train.mean()
sigma2 = X_train.var()
```

```python
def compute_log_prob(X, mu, sigma2):
    log_p = np.zeros(len(X))
    
    for col in X.columns:
        log_p += -0.5 * np.log(2 * np.pi * sigma2[col])
        log_p += - ((X[col] - mu[col])**2) / (2 * sigma2[col])
    
    return log_p
```

```python
log_p_values = compute_log_prob(X, mu, sigma2)
```

```python
from sklearn.model_selection import train_test_split

X_train_full = X[y == 0]   # only normal
X_val, X_test, y_val, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
best_f1 = 0
best_epsilon = 0

for epsilon in np.linspace(min(log_p_values),
                           max(log_p_values),
                           1000):
    
    preds = (log_p_values < epsilon).astype(int)
    
    tp = np.sum((preds == 1) & (y == 1))
    fp = np.sum((preds == 1) & (y == 0))
    fn = np.sum((preds == 0) & (y == 1))
    
    if tp + fp == 0 or tp + fn == 0:
        continue
    
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1 = 2 * precision * recall / (precision + recall)
    
    if f1 > best_f1:
        best_f1 = f1
        best_epsilon = epsilon

print("Best F1:", best_f1)
print("Best Epsilon:", best_epsilon)
```

```python
predictions = (log_p_values < best_epsilon).astype(int)
```

```python
tp = np.sum((predictions == 1) & (y == 1))
fp = np.sum((predictions == 1) & (y == 0))
fn = np.sum((predictions == 0) & (y == 1))

precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * precision * recall / (precision + recall)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
feature = 'radius_mean'

x_vals = np.linspace(X_train[feature].min(),
                     X_train[feature].max(), 200)

mu_f = mu[feature]
sigma_f = np.sqrt(sigma2[feature])

gaussian_curve = (1 / (np.sqrt(2*np.pi)*sigma_f)) * \
                 np.exp(-((x_vals - mu_f)**2)/(2*sigma_f**2))

plt.figure(figsize=(8,5))
plt.hist(X_train[feature], bins=30, density=True, alpha=0.6)
plt.plot(x_vals, gaussian_curve, color='red')
plt.title("Gaussian Fit for radius_mean")
plt.show()
```

```python
epsilons = np.linspace(min(log_p_values),
                       max(log_p_values), 200)

f1_scores = []

for epsilon in epsilons:
    preds = (log_p_values < epsilon).astype(int)
    
    tp = np.sum((preds == 1) & (y == 1))
    fp = np.sum((preds == 1) & (y == 0))
    fn = np.sum((preds == 0) & (y == 1))
    
    if tp + fp == 0 or tp + fn == 0:
        f1_scores.append(0)
        continue
    
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1 = 2 * precision * recall / (precision + recall)
    
    f1_scores.append(f1)

plt.plot(epsilons, f1_scores)
plt.title("F1 Score vs Epsilon")
plt.xlabel("Epsilon")
plt.ylabel("F1 Score")
plt.show()
```
